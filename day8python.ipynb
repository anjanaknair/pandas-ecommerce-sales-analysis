```python
# ================================================================
#       E-COMMERCE SALES DATA ANALYSIS USING PANDAS
#       Internship Assignment - Day 8
# ================================================================

# -------------------- 1. IMPORT LIBRARIES ------------------------

import pandas as pd
import numpy as np

print("=" * 70)
print("           E-COMMERCE SALES DATA ANALYSIS")
print("=" * 70)


# -------------------- 2. LOAD DATASET ----------------------------

from google.colab import files
import pandas as pd

# Upload the CSV file
uploaded = files.upload()

# Get the uploaded file name automatically
file_name = next(iter(uploaded))

# Load the dataset
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File name:", file_name)
print("Dataset shape:", df.shape)

df.head()

print("\n1. DATASET LOADED SUCCESSFULLY")
print("-" * 50)
print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])


# -------------------- 3. DATASET OVERVIEW ------------------------

print("\n2. FIRST FIVE RECORDS")
print("-" * 50)
print(df.head())


print("\n3. LAST FIVE RECORDS")
print("-" * 50)
print(df.tail())


print("\n4. DATASET SHAPE")
print("-" * 50)
print("Shape:", df.shape)


print("\n5. COLUMN NAMES")
print("-" * 50)
print(df.columns.tolist())


print("\n6. DATA TYPES")
print("-" * 50)
print(df.dtypes)


# -------------------- 4. DATA QUALITY CHECK ----------------------

print("\n7. MISSING VALUE CHECK")
print("-" * 50)
print(df.isnull().sum())


print("\n8. DUPLICATE RECORD CHECK")
print("-" * 50)
print("Number of duplicate rows:", df.duplicated().sum())


# -------------------- 5. DATA TYPE CONVERSION --------------------

# Convert Order_Date into datetime format
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("\n9. DATA TYPES AFTER CONVERSION")
print("-" * 50)
print(df.dtypes)


# -------------------- 6. DESCRIPTIVE STATISTICS ------------------

print("\n10. DESCRIPTIVE STATISTICS")
print("-" * 50)
print(df.describe())


# ================================================================
#                    DATA SELECTION
# ================================================================

print("\n" + "=" * 70)
print("                    DATA SELECTION")
print("=" * 70)


# Select a single column
print("\n11. SELECTING TOTAL SALES COLUMN")
print("-" * 50)
print(df["Total_Sales"].head())


# Select multiple columns
print("\n12. SELECTING MULTIPLE COLUMNS")
print("-" * 50)

selected_data = df[
    [
        "Order_ID",
        "Customer",
        "Product",
        "Category",
        "Quantity",
        "Total_Sales"
    ]
]

print(selected_data.head())


# Select rows using iloc
print("\n13. SELECTING ROWS USING iloc")
print("-" * 50)
print(df.iloc[0:5, 0:6])


# ================================================================
#                    DATA FILTERING
# ================================================================

print("\n" + "=" * 70)
print("                     DATA FILTERING")
print("=" * 70)


# Orders with sales greater than 5000
print("\n14. ORDERS WITH SALES GREATER THAN ₹5000")
print("-" * 50)

high_value_orders = df[df["Total_Sales"] > 5000]

print(high_value_orders[
    [
        "Order_ID",
        "Customer",
        "Product",
        "Category",
        "Total_Sales"
    ]
])


# Orders with quantity >= 4
print("\n15. ORDERS WITH QUANTITY GREATER THAN OR EQUAL TO 4")
print("-" * 50)

high_quantity_orders = df[df["Quantity"] >= 4]

print(high_quantity_orders[
    [
        "Order_ID",
        "Product",
        "Quantity",
        "Total_Sales"
    ]
])


# Electronics category
print("\n16. ELECTRONICS CATEGORY ORDERS")
print("-" * 50)

electronics_orders = df[df["Category"] == "Electronics"]

print(electronics_orders[
    [
        "Order_ID",
        "Product",
        "Category",
        "Total_Sales"
    ]
].head(10))


# Multiple conditions
print("\n17. ELECTRONICS ORDERS WITH SALES GREATER THAN ₹5000")
print("-" * 50)

electronics_high_sales = df[
    (df["Category"] == "Electronics") &
    (df["Total_Sales"] > 5000)
]

print(electronics_high_sales[
    [
        "Order_ID",
        "Product",
        "Category",
        "Total_Sales"
    ]
])


# High-rated orders
print("\n18. ORDERS WITH RATING 4.5 OR ABOVE")
print("-" * 50)

high_rated_orders = df[df["Rating"] >= 4.5]

print(high_rated_orders[
    [
        "Order_ID",
        "Customer",
        "Product",
        "Rating",
        "Total_Sales"
    ]
].head(10))


# ================================================================
#                     SORTING OPERATIONS
# ================================================================

print("\n" + "=" * 70)
print("                    SORTING OPERATIONS")
print("=" * 70)


# Highest sales
print("\n19. TOP 10 ORDERS BY SALES")
print("-" * 50)

top_10_orders = df.sort_values(
    by="Total_Sales",
    ascending=False
).head(10)

print(top_10_orders[
    [
        "Order_ID",
        "Customer",
        "City",
        "Category",
        "Product",
        "Total_Sales"
    ]
])


# Lowest sales
print("\n20. 10 LOWEST SALES ORDERS")
print("-" * 50)

lowest_sales = df.sort_values(
    by="Total_Sales",
    ascending=True
).head(10)

print(lowest_sales[
    [
        "Order_ID",
        "Customer",
        "Product",
        "Category",
        "Total_Sales"
    ]
])


# ================================================================
#                   OVERALL SALES ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                  OVERALL SALES ANALYSIS")
print("=" * 70)


total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print("\n21. KEY SALES STATISTICS")
print("-" * 50)

print(f"Total Sales          : ₹{total_sales:,.2f}")
print(f"Average Sales        : ₹{average_sales:,.2f}")
print(f"Highest Sale         : ₹{highest_sales:,.2f}")
print(f"Lowest Sale          : ₹{lowest_sales:,.2f}")
print(f"Total Quantity Sold  : {total_quantity}")
print(f"Number of Orders     : {number_of_orders}")


# ================================================================
#                  CATEGORY-WISE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                  CATEGORY-WISE ANALYSIS")
print("=" * 70)


category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Highest_Sale=("Total_Sales", "max"),
    Lowest_Sale=("Total_Sales", "min"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("\n22. SALES PERFORMANCE BY CATEGORY")
print("-" * 50)
print(category_analysis)


top_category = category_analysis.index[0]

print("\nTop-performing category:", top_category)
print(
    f"Total sales of {top_category}: "
    f"₹{category_analysis.loc[top_category, 'Total_Sales']:,.2f}"
)


# ================================================================
#                     CITY-WISE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                     CITY-WISE ANALYSIS")
print("=" * 70)


city_analysis = df.groupby("City").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Highest_Sale=("Total_Sales", "max"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("\n23. SALES PERFORMANCE BY CITY")
print("-" * 50)
print(city_analysis)


top_city = city_analysis.index[0]

print("\nTop-performing city:", top_city)
print(
    f"Total sales of {top_city}: "
    f"₹{city_analysis.loc[top_city, 'Total_Sales']:,.2f}"
)


# ================================================================
#                    PRODUCT-WISE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                   PRODUCT-WISE ANALYSIS")
print("=" * 70)


product_analysis = df.groupby("Product").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Highest_Sale=("Total_Sales", "max"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("\n24. TOP 10 PRODUCTS BY TOTAL SALES")
print("-" * 50)
print(product_analysis.head(10))


top_product = product_analysis.index[0]

print("\nTop-performing product:", top_product)
print(
    f"Total sales of {top_product}: "
    f"₹{product_analysis.loc[top_product, 'Total_Sales']:,.2f}"
)


# Top products by quantity
print("\n25. TOP 10 PRODUCTS BY QUANTITY SOLD")
print("-" * 50)

quantity_by_product = df.groupby("Product")["Quantity"].sum().sort_values(
    ascending=False
)

print(quantity_by_product.head(10))


# ================================================================
#                 PAYMENT METHOD ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                PAYMENT METHOD ANALYSIS")
print("=" * 70)


payment_analysis = df.groupby("Payment_Method").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("\n26. SALES PERFORMANCE BY PAYMENT METHOD")
print("-" * 50)
print(payment_analysis)


# Most frequently used payment method
payment_counts = df["Payment_Method"].value_counts()

most_used_payment = payment_counts.index[0]

print("\n27. MOST USED PAYMENT METHOD")
print("-" * 50)
print("Payment Method:", most_used_payment)
print("Number of Orders:", payment_counts.iloc[0])


# ================================================================
#                    REGION-WISE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                    REGION-WISE ANALYSIS")
print("=" * 70)


region_analysis = df.groupby("Region").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("\n28. SALES PERFORMANCE BY REGION")
print("-" * 50)
print(region_analysis)


# ================================================================
#                  SPECIFIC ORDER ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("                  SPECIFIC ORDER ANALYSIS")
print("=" * 70)


# Orders above ₹10,000
print("\n29. ORDERS WITH SALES ABOVE ₹10,000")
print("-" * 50)

orders_above_10000 = df[df["Total_Sales"] > 10000].sort_values(
    by="Total_Sales",
    ascending=False
)

print(orders_above_10000[
    [
        "Order_ID",
        "Customer",
        "City",
        "Category",
        "Product",
        "Quantity",
        "Total_Sales",
        "Payment_Method"
    ]
])


# Highest individual order
print("\n30. HIGHEST-VALUE ORDER")
print("-" * 50)

highest_order = df.loc[df["Total_Sales"].idxmax()]

print("Order ID      :", highest_order["Order_ID"])
print("Customer      :", highest_order["Customer"])
print("City          :", highest_order["City"])
print("Category      :", highest_order["Category"])
print("Product       :", highest_order["Product"])
print("Quantity      :", highest_order["Quantity"])
print(f"Total Sales   : ₹{highest_order['Total_Sales']:,.2f}")
print("Payment Method:", highest_order["Payment_Method"])


# Lowest individual order
print("\n31. LOWEST-VALUE ORDER")
print("-" * 50)

lowest_order = df.loc[df["Total_Sales"].idxmin()]

print("Order ID      :", lowest_order["Order_ID"])
print("Customer      :", lowest_order["Customer"])
print("City          :", lowest_order["City"])
print("Category      :", lowest_order["Category"])
print("Product       :", lowest_order["Product"])
print("Quantity      :", lowest_order["Quantity"])
print(f"Total Sales   : ₹{lowest_order['Total_Sales']:,.2f}")
print("Payment Method:", lowest_order["Payment_Method"])


# ================================================================
#               CATEGORY AVERAGE SALES ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("               CATEGORY AVERAGE SALES ANALYSIS")
print("=" * 70)


category_average = df.groupby("Category")["Total_Sales"].mean().sort_values(
    ascending=False
)

print("\n32. CATEGORIES RANKED BY AVERAGE SALES")
print("-" * 50)
print(category_average)

highest_average_category = category_average.index[0]

print(
    "\nCategory with highest average sales:",
    highest_average_category
)


# ================================================================
#                  PAYMENT METHOD USAGE
# ================================================================

print("\n" + "=" * 70)
print("                 PAYMENT METHOD USAGE")
print("=" * 70)


print("\n33. NUMBER OF ORDERS BY PAYMENT METHOD")
print("-" * 50)
print(payment_counts)


# ================================================================
#                  MEANINGFUL OBSERVATIONS
# ================================================================

print("\n" + "=" * 70)
print("                  MEANINGFUL OBSERVATIONS")
print("=" * 70)


print(f"""
1. The dataset contains {number_of_orders} orders across
   {df['Category'].nunique()} product categories and
   {df['City'].nunique()} cities.

2. The total sales generated from all orders is
   ₹{total_sales:,.2f}, with an average order value of
   ₹{average_sales:,.2f}.

3. The highest individual order generated
   ₹{highest_sales:,.2f}, while the lowest order generated
   ₹{lowest_sales:,.2f}.

4. A total of {total_quantity} units were sold across all
   {number_of_orders} orders.

5. {top_category} is the top-performing category based on
   total sales, generating
   ₹{category_analysis.loc[top_category, 'Total_Sales']:,.2f}.

6. {top_city} is the highest-performing city based on total
   sales, contributing
   ₹{city_analysis.loc[top_city, 'Total_Sales']:,.2f}.

7. {top_product} is the top-performing product based on
   total sales, generating
   ₹{product_analysis.loc[top_product, 'Total_Sales']:,.2f}.

8. {most_used_payment} is the most frequently used payment
   method, accounting for {payment_counts.iloc[0]} orders.

9. Filtering operations helped identify high-value orders,
   including orders with sales above ₹10,000.

10. Groupby() and aggregation functions made it possible to
    compare sales performance across categories, cities,
    products, regions, and payment methods.
""")


# ================================================================
#                         CONCLUSION
# ================================================================

print("\n" + "=" * 70)
print("                         CONCLUSION")
print("=" * 70)

print("""
This analysis demonstrates how Python and Pandas can be used
to analyze an e-commerce sales dataset and extract meaningful
business insights.

The analysis covered data loading, data selection, filtering,
sorting, groupby(), and aggregation operations. Sales were
analyzed based on categories, cities, products, regions, and
payment methods.

Important statistics such as total sales, average sales,
highest and lowest sales, total quantity sold, and number of
orders were calculated. The analysis also identified
top-performing products and categories and specific
high-value orders.

Therefore, Pandas provides an efficient and powerful way to
transform raw e-commerce data into useful information for
business analysis and decision-making.
""")


print("=" * 70)
print("             E-COMMERCE DATA ANALYSIS COMPLETED")
print("=" * 70)
```
